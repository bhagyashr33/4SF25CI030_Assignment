1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.



```python
import pandas as pd
import numpy as np

df = pd.DataFrame(
    np.random.randint(1, 11, (5, 3)),
    columns=["A", "B", "C"]
)

df["Product"] = df["A"] * df["B"]

print(df)
```

2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

```python
df = pd.DataFrame(
    np.random.randint(1, 11, (4, 3)),
    columns=["A", "B", "C"]
)

row_sum = df.sum(axis=1)
column_sum = df.sum(axis=0)

print(df)
print("Row-wise sum:")
print(row_sum)
print("Column-wise sum:")
print(column_sum)
```

3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.


```python
df = pd.DataFrame(
    np.random.randint(1, 11, (5, 3)),
    columns=["A", "B", "C"]
)

df.iloc[1, 0] = np.nan
df.iloc[3, 1] = np.nan
df.iloc[2, 2] = np.nan

df = df.fillna(df.mean())

print(df)
```

4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.



```python
df = pd.DataFrame(
    np.random.randint(1, 11, (6, 4)),
    columns=["A", "B", "C", "D"]
)

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 1] = np.nan

print("Before:")
print(df)

df = df.dropna()

print("After:")
print(df)
```

5. Data Aggregation
Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.


```python
import pandas as pd

df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Value': [10, 20, 30, 40, 50, 60]
})

print(df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
import pandas as pd

df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Bag', 'Pencil', 'Bottle'],
    'Category': ['A', 'B', 'A', 'B', 'A'],
    'Sales': [100, 200, 150, 120, 180]
})

print(df.groupby('Category')['Sales'].sum())
```

7. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 85]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
import pandas as pd

df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'Y'],
    'Value': [10, 20, 30, 40]
})

df = df.set_index(['Category', 'SubCategory'])

print(df.groupby(['Category', 'SubCategory'])['Value'].sum())
```

10. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [6, 7, 8, 9, 10],
    'C': [11, 12, 13, 14, 15]
})

print(df.map(lambda x: x * 2))
```

11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.


```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3, 4, 5, 6],
    'B': [2, 3, 4, 5, 6, 7],
    'C': [3, 4, 5, 6, 7, 8]
})

df['Sum'] = df.apply(lambda x: x.sum(), axis=1)

print(df)
```
