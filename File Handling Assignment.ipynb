2. Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
with open('data.txt', 'r') as file:
    content = file.read()
content = content.replace('data', 'data(replaced)')
with open('data.txt', 'w') as file:
    file.write(content)
print(content)
```

3. Reading a File in Reverse
Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.


```python
def reverse_lines():
    with open("reverse.txt", "r") as file:
        lines = file.readlines()

    for line in lines:
        print(line.strip()[::-1])
```

4. Counting Lines, Words, and Characters
Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.


```python
def file_stats():
    with open("stats.txt", "r") as file:
        lines = file.readlines()

    line_count = len(lines)
    word_count = 0
    char_count = 0

    for line in lines:
        word_count += len(line.split())
        char_count += len(line)

    return line_count, word_count, char_count
print(file_stats())
```
